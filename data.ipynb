```python
import pandas as pd
```

```python
# 读取原始 CSV 文件
data = pd.read_csv("customers.csv")
```

```python
# 删除 ' weeks' 字符并转换为数值
data['weeks_since_signup'] = data['weeks_since_signup'].replace(r' weeks', '', regex=True).astype(float)
data['weeks_since_last_purchase'] = data['weeks_since_last_purchase'].replace(r' weeks', '', regex=True).astype(float)

# 删除 '$' 字符并转换为数值
data['avg_AddOnpurchase_value'] = data['avg_AddOnpurchase_value'].replace(r'[$]', '', regex=True).astype(float)

# 保存为新文件（可选）
data.to_csv("customers_cleaned.csv", index=False)
```
